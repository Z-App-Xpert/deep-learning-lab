```python
import tensorflow as tf

print("TensorFlow:", tf.__version__)
print("GPUs:", tf.config.list_physical_devices('GPU'))
```

```python

x = tf.constant([[-2.0], [-1.0], [0.0], [1.0], [2.0]])

print("Input")
print(x)

print("ReLU")
print(tf.nn.relu(x))

print("Sigmoid")
print(tf.nn.sigmoid(x))

print("Tanh")
print(tf.nn.tanh(x))
```
