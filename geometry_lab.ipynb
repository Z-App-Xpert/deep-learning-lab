Purpose:  Show one point in 2D space

```python
import numpy as np

x = np.array([1,1])
print(x)
```

Purpose of below cell is to move the point

```python
translation = np.array([3,2])

new_x = x + translation
print(new_x)
```

The purpose of below Cell is to stretch one dimension

```python
S = np.array([
    [2,0],
    [0,1]
])

scaled_x = S @ x
print(scaled_x)
```

Rotation:
The purpose of the below cell is to rotate the vector

```python
R = np.array([
    [0,-1],
    [1,0]
])

rotated_x = R @ x
print(rotated_x)
```

Dense Layer idea's purpose is the neural-network core idea. See cell below.

```python
W = np.array([
    [2,0],
    [0,1]
])

b = np.array([3,2])

y =  W @ x + b

print(y)
```

Add ReLU
Purpose of ReLU is to create the "fold."  Without activation funtions like ReLU that are nonlinear, then stacekd layers collapse into one linear transformation.

```python
def relu(v):
    return np.maximum(0,v)

W = np.array([
    [-2,0],
    [0,1]
])

b = np.array([0,0])

y = W @ x + b

print("Before ReLU:", y)
print("After ReLU:", relu(y))
```

Big Takeaway:

Vector
↓
Wx + b # stretch / rotate / translate
↓
ReLU # fold / bend / clip
↓
New vector

## Tiny Training Loop: One Weight Learning

This example shows the basic learning process of a neural network using only one input, one weight, and one target value.

We want the model to learn:

**x = 2 → y = 6**

The model uses:

**y_pred = w × x**

The correct weight would eventually be:

**w = 3**

because:

**3 × 2 = 6**

However, we start the weight at `1.0` and let the model improve it through training.

### During each training step

1. **Prediction**

   Calculate the model's current prediction:

   `y_pred = w * x`

2. **Loss**

   Measure how far the prediction is from the correct answer:

   `loss = (y_pred - y_true) ** 2`

3. **Gradient**

   Calculate how changing the weight would affect the loss:

   `gradient = 2 * (y_pred - y_true) * x`

   The gradient tells us the direction the weight should move to reduce the loss.

4. **Gradient Descent**

   Update the weight:

   `w = w - learning_rate * gradient`

The `learning_rate` controls how large each weight adjustment is.

### Training loop

**prediction → loss → gradient → weight update → repeat**

A real neural network follows this same basic process, but with many more weights.

```python
x = 2
y_true = 6

w = 1.0
learning_rate = 0.1

for step in range(5):
    y_pred = w * x

    loss = (y_pred - y_true) ** 2

    gradient = 2 * (y_pred - y_true) * x

    w = w - learning_rate * gradient

    print(step, w, loss)
```

Next we add the bias term.

So far our tiny model was:

y_pred = w * x

That works for relationships that pass through zero. But many real relationships do not.

Suppose the true relationship is:

y = 3x + 2

If x = 2, then:

y = 3(2) + 2 = 8

Now the model needs two learnable values:

w = weight
b = bias

So the model becomes:

y_pred = w * x + b

This is the same core idea we saw in a Dense layer:

Y = W @ X + b

The bias lets the model shift the line up or down instead of forcing it through zero.

A tiny training example looks like this:

The important new idea is this:

w controls how strongly the input affects the output.

b gives the neuron a baseline shift.

You can picture it as:

weight = tilt

bias = vertical shift

So if the model is a line:

y = wx + b

then changing w changes the slope, while changing b moves the whole line up or down.

And both are learned using the same process:

prediction → loss → gradients → update w and b

This is now very close to what a real Keras Dense layer is doing internally.

```python
x = 2
y_true = 8

w = 1.0
b = 0.0
learning_rate = 0.1

for step in range(5):
    y_pred = w * x + b

    loss = (y_pred - y_true) ** 2

    gradient_w = 2 * (y_pred - y_true) * x
    gradient_b = 2 * (y_pred - y_true)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b

    print(step, w, b, loss)
```

The important new idea is this:

w controls how strongly the input affects the output.

b gives the neuron a baseline shift.

You can picture it as:

weight = tilt

bias = vertical shift

y = wx + b

then changing w changes the slope, while changing b moves the whole line up or down.

prediction → loss → gradients → update w and b
